3. Порахуйте значення R^2 та MSE для оцінки якості кожної моделі (теж доцільно побудувати графік залежності R^2 або MSE від порядку поліному моделі).
4. Оберіть найкращу модель, висновок обгрунтуйте.


```python
from sklearn.metrics import mean_squared_error, r2_score

# Для зручності обчислення
x = df['population']
y = df['co2_emission']

# Порядки поліномів, які будемо тестувати
degrees = [1, 2, 3, 4, 5]

# Словники для збереження метрик
models = {}
mse_values = []
r2_values = []

# Побудова моделей
for d in degrees:
    coeffs = np.polyfit(x, y, d)
    model = np.poly1d(coeffs)
    y_pred = model(x)

    mse = mean_squared_error(y, y_pred)
    r2 = r2_score(y, y_pred)

    models[d] = model
    mse_values.append(mse)
    r2_values.append(r2)

    print(f"Порядок: {d}")
    print(f"R² = {r2:.4f}, MSE = {mse:.2f}")
    print(model, "\n")
```

```python
plt.figure(figsize=(12, 8))
plt.scatter(x, y, color='black', label='Дані', alpha=0.5)

x_vals = np.linspace(x.min(), x.max(), 500)

# Кольори для графіків
colors = ['r', 'g', 'b', 'orange', 'purple']

for i, d in enumerate(degrees):
    y_vals = models[d](x_vals)
    plt.plot(x_vals, y_vals, label=f'Степінь {d}', color=colors[i])

plt.xlabel("Населення (population)")
plt.ylabel("Викиди CO2")
plt.title("Поліноміальні моделі різного порядку")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(14, 5))

# R² графік
plt.subplot(1, 2, 1)
plt.plot(degrees, r2_values, marker='o', linestyle='-', color='green')
plt.title('Залежність R² від порядку поліному')
plt.xlabel('Порядок поліному')
plt.ylabel('R²')
plt.grid(True)

# MSE графік
plt.subplot(1, 2, 2)
plt.plot(degrees, mse_values, marker='o', linestyle='-', color='red')
plt.title('Залежність MSE від порядку поліному')
plt.xlabel('Порядок поліному')
plt.ylabel('MSE')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
best_degree = degrees[np.argmax(r2_values)]
print(f"Найкраща модель — поліном порядку {best_degree}")
```
